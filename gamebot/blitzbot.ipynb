# 介紹

本程式使用 pyautogui 來完成一個 Blitz 的遊戲機器人 

```python
import pyautogui as pyg 
pyg.sleep(2)
basex=225
basey=395
```

```python
gap = 40  # 請依畫面中寶石的實際間距調整
screen = pyg.screenshot()
for row in range(8):
    for column in range(8):
        x = basex + column * gap
        y = basey + row * gap
        color = screen.getpixel((x, y))
        print(f"第 {row + 1} 列、第 {column + 1} 欄：{color}")
```
