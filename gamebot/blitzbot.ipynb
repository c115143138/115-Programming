# 介紹

本程式使用 pyautogui 來完成一個 Blitz 的遊戲機器人 

```python
import pyautogui as pyg 
pyg.sleep(2)
basex=120
basey=304
gap=40
```

```python
gap = 40  # 請依畫面中寶石的實際間距調整
screen = pyg.screenshot()
for row in range(8):
    for column in range(8):
        x = basex + column * gap
        y = basey + row * gap
        color = screen.getpixel((x, y))
        print(f"第 {row + 1} 列、第 {column + 1} 欄：{color}")
```

```python
def gem_position(r, c):
    """回傳第 r 列、第 c 欄寶石的螢幕座標（r、c 從 0 開始）。"""
    x = basex + c * gap
    y = basey + r * gap
    return (x, y)

# 範例：第 1 列、第 1 欄（左上角）
gem_position(0, 0)
```

```python
print(gem_position(0, 0))
print(gem_position(0, 1))
print(gem_position(1, 0))
```

```python
def get_gem_color(r, c, im):
    """回傳截圖 im 中第 r 列、第 c 欄寶石的顏色（r、c 從 0 開始）。"""
    x, y = gem_position(r, c)
    return im.getpixel((x, y))
```

```python
im = pyg.screenshot()
for i in range(8):
    x,y =  gem_position(0, i)
    pyg.click(x,y, duration=1)
    print(get_gem_color(0, i, im))
```

```python
def can_move_left(r, c, im):
    """若將 (r, c) 與左側寶石交換後形成三連線，回傳 True。r、c 從 0 開始。"""
    if not (0 <= r < 8 and 1 <= c < 8):
        return False

    # 取得交換後的顏色；其他格子仍從截圖讀取。
    def color_at(row, column):
        if (row, column) == (r, c):
            return get_gem_color(r, c - 1, im)
        if (row, column) == (r, c - 1):
            return get_gem_color(r, c, im)
        return get_gem_color(row, column, im)

    def makes_match(row, column, dr, dc):
        color = color_at(row, column)
        count = 1
        for direction in (-1, 1):
            next_row = row + direction * dr
            next_column = column + direction * dc
            while (0 <= next_row < 8 and 0 <= next_column < 8
                   and color_at(next_row, next_column) == color):
                count += 1
                next_row += direction * dr
                next_column += direction * dc
        return count >= 3

    # 新形成的連線必定經過交換的其中一格。
    for row, column in ((r, c), (r, c - 1)):
        if makes_match(row, column, 0, 1) or makes_match(row, column, 1, 0):
            return True
    return False
```

```python
im = pyg.screenshot()
print(can_move_left(1, 3, im))
```

```python
def _can_move_direction(r, c, im, dr, dc):
    """模擬與指定方向的相鄰寶石交換，判斷是否形成三連線。"""
    next_r = r + dr
    next_c = c + dc
    if not (0 <= r < 8 and 0 <= c < 8
            and 0 <= next_r < 8 and 0 <= next_c < 8):
        return False

    def color_at(row, column):
        if (row, column) == (r, c):
            return get_gem_color(next_r, next_c, im)
        if (row, column) == (next_r, next_c):
            return get_gem_color(r, c, im)
        return get_gem_color(row, column, im)

    def makes_match(row, column, row_step, column_step):
        color = color_at(row, column)
        count = 1
        for direction in (-1, 1):
            check_r = row + direction * row_step
            check_c = column + direction * column_step
            while (0 <= check_r < 8 and 0 <= check_c < 8
                   and color_at(check_r, check_c) == color):
                count += 1
                check_r += direction * row_step
                check_c += direction * column_step
        return count >= 3

    for check_r, check_c in ((r, c), (next_r, next_c)):
        if makes_match(check_r, check_c, 0, 1):
            return True
        if makes_match(check_r, check_c, 1, 0):
            return True
    return False

def can_move_right(r, c, im):
    """判斷 (r, c) 是否可向右移動；r、c 從 0 開始。"""
    return _can_move_direction(r, c, im, 0, 1)

def can_move_up(r, c, im):
    """判斷 (r, c) 是否可向上移動；r、c 從 0 開始。"""
    return _can_move_direction(r, c, im, -1, 0)

def can_move_down(r, c, im):
    """判斷 (r, c) 是否可向下移動；r、c 從 0 開始。"""
    return _can_move_direction(r, c, im, 1, 0)
```

```python
import time

start_time = time.time()
im = pyg.screenshot()
directions = (
    (can_move_left, 0, -1),
    (can_move_right, 0, 1),
    (can_move_up, -1, 0),
    (can_move_down, 1, 0),
)

while time.time() - start_time < 60:
    moved = False
    for r in range(8):
        if time.time() - start_time >= 60:
            break
        for c in range(8):
            if time.time() - start_time >= 60:
                break
            for can_move, dr, dc in directions:
                if can_move(r, c, im):
                    x, y = gem_position(r, c)
                    target_x, target_y = gem_position(r + dr, c + dc)
                    pyg.moveTo(x, y)
                    pyg.dragTo(target_x, target_y, duration=0.15, button="left")
                    pyg.sleep(0.2)
                    im = pyg.screenshot()
                    moved = True
                    break
    if not moved:
        im = pyg.screenshot()
        pyg.sleep(0.05)
```
